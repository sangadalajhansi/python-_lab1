```python
#for loop
#1.print 1ton odd numbers
n=int(input())
for i in range(1,n+1,2):
  print(i,end=" ")
```

```python
#2.print multiplication table of a given number 2 * 1 = 2
n=int(input())
for i in range(1,10):
  print(f"{n} * {i} + {n*i}")
```

```python
#3.check whether the given num is prime or not
n=int(input())
for i in range(2,n):
  print("it is not prime")
  break
else :
 print("yes it is prime")
```

```python
#4.print prime numbers between given 2 numbers
a=int(input())
b=int(input())
for i in range(a,b+1):
 for j in range(2,i):
   if(i%j==0):
    break
 else:
  print(i,end=" ")
```

```python
#5.print the sum of the digits of the given numbers
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```

```python
#6.print 1 to n numbers 10 times
for i in range(1,11):
  for j in range(1,11):
    print(i,end=" ")
print("\n")
```
