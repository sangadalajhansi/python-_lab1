```python
#check whether given number is palindrome or not using functions
def palindrome (n):
  original = n
  reverse = 0
  while n>0 :
    digit = n%10
    reverse =reverse*10+digit
    n=n//10
  if original ==reverse:
     return True
  else:
     return False
num=int(input("enter a number:"))
if palindrome(num):
    print("palindrome")
else:
    print("not palindrome")
```

```python
def reverse_str(s):
  rev=""
  for i in range(len(s)-1,-1,-1):
    rev=rev+s[i]
    return rev
print(reverse_str("janu"))
```

```python
#function that returns multiple
def multiple_values (*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4,5)
print(f"sum is {res_sum},count is{res_count},the average is{res_avg}")
```
