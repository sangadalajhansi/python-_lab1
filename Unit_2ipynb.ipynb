```python
L1=[1,2,3,55,66,88,99.0,"hello","name"]
L2=[100,220,440]
print(L1+L2) #concatenation
print(L2*3) #repetition
print(L1[5]) #indexing , print element at 5th index
print(100 in L2) #check 100 present in list L2 or not
print(L1[2:7]) #slicing, prints elements from 2nd index to 6th index
print(L1[::-1]) #print elements in reverse order
print(L1[-7:-1])
```
